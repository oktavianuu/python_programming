````python
def happy_new_year(wishes = True):
    print("Three...")
    print("Two...")
    print("One...")
    if not wishes:
        return

    print("Happy New Year!")

````

````python
happy_new_year()
````

````python
happy_new_year(False)
````

Notice the difference of the output above? It means that if `wishes` is set to be false, then the execution will be stopped. If not then it will executes all of those line until the last line. This kind of return act as a control, meaning that it can stop the program execution at certain point. Another variant of return functioned to return values. 
```python
def function():
    return expression
```

````python
def boring_function():
    return 123

x = boring_function()

print("The boring_function has returned its result. It's:", x)

````

The `return` keyword transports the value to the invoked function. The result may be freely used, e.g, assigned to a varible or be manipulated. The result may be completely ignored and lost without a trace. 

````python
def boring_function():
    print("'Boredom Mode' ON.")
    return 123

print("This lesson is interesting!")
boring_function()
print("This lesson is boring...")
````

Notice that the value is not displayed because it is ignored and it is okay. It won't cause any error et all.

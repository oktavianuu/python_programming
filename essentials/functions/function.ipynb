## Positional Argument

```python
def introduction(first_name, last_name):
    print("Hello, my name is", first_name, last_name)

introduction("Skywalker", "Luke")
introduction("Quick", "Jesse")
introduction("Kent", "Clark")

```

Positional argument means that the argument is passed to the function based on its position. 

### Keyword Argument Passing
Just like its name, the arguments are not passed based on their positions, instead it uses keyword or the parameters' name. 

```python
def introduction(first_name, last_name):
    print("Hello, my name is", first_name, last_name)

introduction(first_name = "James", last_name = "Bond")
introduction(last_name = "Skywalker", first_name = "Luke")
```

```python
def introduction(first_name, last_name):
    print("Hello, my name is", first_name, last_name)

introduction(surname="Skywalker", first_name="Luke")

```

The error above is caused by the arguments name in the function calll do not matched.

### Mixing positional and keyword arguments
We can mix both but with one absolute rule that we must put the positional argument before keyword argument. 

```python
def adding(a, b, c):
    print(a, '+', b, '+', c, '=', a + b + c)

adding(1, 2, 3) # positional argument passing
```

```python
adding(1, c=3, b=2) # mix both with correct rule
```

```python
adding(b=2, 1, c=3) # violates the rule that the positional argument must be passed before the keyword arg
```

```python
adding(1, 2, c=3) # this will correct since the first two vaues is passed as positional argument.
```

### Default Argument Values
There will be times where we will use an argument value more often. This makes us want to set that value as default so we do not have to change or set it everytime we call the function. 

```python
def introduction(first_name, last_name='Doe'):
    print('Hello, my name is', first_name, last_name)
```

```python
introduction('John') # by calling the function this way, the default value will always be used even though we not typing it.
```

```python
introduction('John', 'Cena') # this is also valid. We replace the default value with 'Cena'
```

```python
introduction(first_name='John')
```

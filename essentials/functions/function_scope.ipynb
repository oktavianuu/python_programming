```python
def scope_test():
    x = 123

scope_test()
print(x)
```

Look at what happen above? The program throw an error since `x` inside the function is inaccessible outside the function.

```python
def my_function():
    print("Do I know a variable?", var)

var = 1
my_function()
print(var)
```

And look at the result produced by the function above! the var outside the function is accessible inside the function and the function recognize it. 

**Formal:  a variable existing outside a function has scope inside the function's body.**

```python
def my_function():
    var = 2
    print("Do I know that variable?", var)


var = 1
my_function()
print(var)
```

The var inside a function will shadow the var outside the function.
